```python
# 导入模块
import os 
from pathlib import Path
import pandas as pd
import numpy as np
```

```python
# 逐日读取工作簿，次序由文件名中的日期决定。
filenames = sorted(Path('data').glob('*.xlsx'))
filenames
```

```python
# 筛选掉 “合计”无用项，并设置target_col存储目标项
f = filenames[0]
# 前五行是无用数据
file = pd.read_excel(f, skiprows = 5, skipfooter = 3)
tarr = file.values
print(tarr[3])
test = tarr[0]
target_col = []
for i in range(len(test)):
    tmp = test[i]
    if tmp != '合计':
        target_col.append(i)
print(target_col)
```

```python
# 获取车站名和车站编号
nfile = pd.read_excel(f, skiprows = 5, skipfooter = 3, usecols = target_col)
arrt = nfile.values
stations_name = []
stations_index = []
for i in range(2,len(arrt[0])):
    stations_index.append(i)
    stations_name.append(arrt[0][i])
print(stations_name)
print(stations_index)
```

```python
def read_daily_flow(path):
    frame = pd.read_excel(path, skiprows=5, skipfooter=3, usecols=target_col)
    if frame.iloc[0, 2:].tolist() != stations_name:
        raise ValueError(f'车站顺序与首日不一致 {path.name}')
    # 沿用教学样表的前 43 行剔除规则。
    body = frame.iloc[43:]
    inflow = body.loc[body.iloc[:, 1] == '进站'].iloc[:, 2:].T.reset_index(drop=True)
    outflow = body.loc[body.iloc[:, 1] == '出站'].iloc[:, 2:].T.reset_index(drop=True)
    return inflow, outflow
```

```python
day_flows = [read_daily_flow(path) for path in filenames]
df_in = pd.concat([day_in for day_in, _ in day_flows], axis=1, ignore_index=True)
df_out = pd.concat([day_out for _, day_out in day_flows], axis=1, ignore_index=True)
```

```python
output_dir = Path(os.environ.get('COURSE_BASIC_OUTPUT', '输出文件'))
output_dir.mkdir(parents=True, exist_ok=True)
df_in.to_csv(output_dir / 'in_test.csv', index=False)
df_out.to_csv(output_dir / 'out_test.csv', index=False)
print('Pandas daily reports:', df_in.shape, df_out.shape)
```
