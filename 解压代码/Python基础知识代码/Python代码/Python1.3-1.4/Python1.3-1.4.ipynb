```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get(self):
        print(self.name, self.age)
```

```python
class Student(Human):

    def __init__(self, name, age, stuID):
        # 调用父类的实例化方法
#         super(Student, self).__init__
        Human.__init__(self, name, age)
        self.stuID = stuID

    # 重写父类的get方法
    def get(self):
        print(self.name, self.age, self.stuID)
```

```python
student1 = Student('小明', 10, 12)
```

```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get(self):
        print(self.name, self.age)
```

```python
class Student:
    def __init__(self, stuID):
        self.stuID = stuID

    def get(self):
        print(self.stuID)
```

```python
class CollegeStudent(Human, Student):

    def __init__(self, name, age, stuID, CET6Score):
        Human.__init__(self, name, age)
        Student.__init__(self, stuID)
        self.CET6Score = CET6Score

    # 重写父类的get方法
    def get(self):
        print(self.name, self.age, self.stuID, self.CET6Score)
```

```python
student2 = CollegeStudent('小明', 20, 12, 532)
```

```python
class Parent:
   def func(self):
      print("parent")

class Child(Parent):
   def __init__(self):
      super().__init__()  # 调用父类的构造方法

   def func(self):
      super().func()  # 调用父类的其他方法
      print("child")

object = Child()
object.func()
```

```python
class Parent:
    def func(self):
        print("parent")

class Child1(Parent):
    def func(self):
        print("child1")

class Child2(Parent):
    def func(self):
        print("Child2")

object = Parent()
object.func()

object = Child1()
object.func()

object = Child2()
object.func()
```

```python
file = open("./test.txt", "r", encoding="utf-8")
a = file.readline()  # 读取一行
print('a:', a) # My name is Rachel
file.close() 


file = open("./test.txt", "r", encoding="utf-8")
b = file.readlines() # 逐行读取剩余全部行
print('b:', b)  # ['I am 20 years old\n', 'I like reading books\n']
file.close()

file = open("./test.txt", "r", encoding="utf-8")
c = file.read()  # 一次性读取整个文档
print( 'c:', c)  # 打印为空，因为readlines已经读完整个文件了
file.close()  # 读取完毕
```

```python
import os
from pathlib import Path

output_dir = Path(os.environ.get('COURSE_BASIC_OUTPUT', '输出文件'))
output_dir.mkdir(parents=True, exist_ok=True)

with open("./test.txt", "r", encoding="utf-8") as file:
    a = file.readline()  # 读取一行
    print('a:', a) # My name is Rachel
    b = file.readlines() # 逐行读取剩余全部行
    print('b:', b)  # ['I am 20 years old\n', 'I like reading books\n']
    c = file.read()  # 一次性读取整个文档
    print('C:', c)  # 打印为空，因为readlines已经读完整个文件了
    # 此处不需要使用file.close()

with (output_dir / "test2.txt").open("w", encoding="utf-8") as file:
    file.write("My name is Rachel")
```
