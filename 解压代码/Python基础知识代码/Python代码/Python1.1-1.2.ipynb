```python
i = 1
while i < 6:
    i += 1
    if i % 2 > 0:  # 非双数时跳过输出
        continue
    print(i)
# 输出双数2、4、6
```

```python
i = 1
while 1:  # 循环条件为1必定成立
    print(i)
    # 输出1~3
    i += 1
    if i > 3:  # 当i大于3时跳出循环
        break
```

```python
city = ['北京', '上海', '深圳']
for index in range(len(city)):
    print('当前城市 : %s' % city[index])
print("Good bye!")
# 输出结果：
# 当前城市 : 北京
# 当前城市 : 上海
# 当前城市 : 深圳
# Good bye!
```

```python
for num in range(10, 13):   # 迭代 10 到 13 之间的数字
    for i in range(2, num):  # 根据因子迭代
        if num % i == 0:      # 确定第一个因子
            j = num / i        # 计算第二个因子
            print('%d 等于 %d * %d' % (num, i, j))
            break              # 跳出当前循环
    else:                    # 循环的 else 部分
        print('%d 是一个质数' % num)
# 输出结果为：
# 10 等于 2 * 5
# 11 是一个质数
# 12 等于 2 * 6
```

```python
list = [x*x for x in range(3)]
print(list)  # [0, 1, 4]
```

```python
list = [x*x for x in range(3) if x % 2 == 0]
print(list)  # [0, 4]
```

```python
list = [(x, y) for x in range(2) for y in range(2)]
print(list)  # [(0, 0), (0, 1), (1, 0), (1, 1)]
```

```python
list = [(x, y) for x in range(4) if x % 2 == 0 for y in range(4) if y % 2 == 0]
print(list)  # [(0, 0), (0, 2), (2, 0), (2, 2)]
```
