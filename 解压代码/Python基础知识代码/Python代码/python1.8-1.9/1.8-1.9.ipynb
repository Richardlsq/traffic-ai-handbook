```python
import numpy as np
from sklearn import svm
import pandas as pd
import matplotlib.pyplot as plt

# 正类，服从（2，1）的正态分布
rng = np.random.default_rng(42)
p = rng.normal(size=(200, 2))
for i in range(200):
    p[i][0] += 2
    p[i][1] += 2
# 负类，服从（0，1）的正态分布
f = rng.normal(size=(200, 2))

# 将np数组转换成dataframe
df_p = pd.DataFrame(p, columns=['x', 'y'])
# 加上标签z，正类标签为1
df_p['z'] = 1
# 负类标签为0
df_f = pd.DataFrame(f, columns=['x', 'y'])
df_f['z'] = 0

# 将正负类合并成一个dataframe(按行进行合并)
con = pd.concat([df_p,df_f],axis=0)
print(con)
```

```python
from sklearn.model_selection import train_test_split

train_data, test_data = train_test_split(con, test_size=150, stratify=con['z'], random_state=42)
X = train_data[['x', 'y']]
Z = train_data['z']
clf = svm.SVC(kernel='linear').fit(X, Z)
print('SVM test support:', test_data['z'].value_counts().sort_index().to_dict())
print('SVM test accuracy:', clf.score(test_data[['x', 'y']], test_data['z']))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor

# 导入西直门地铁站点15min进站客流
df = pd.read_csv('./xizhimen.csv', encoding="gbk", parse_dates=True)
len(df)
print(df.head())  # 观察数据集，这是一个单变量时间序列

plt.figure(figsize=(15, 5))
plt.grid(True)  # 网格化
plt.plot(df.iloc[:, 0], df.iloc[:, 1], label="XiZhiMen Station")
plt.legend()
plt.show()
```

```python
p_flow = df['p_flow']
history = p_flow.shift(1)
df['pre_date_flow'] = history
df['MA5'] = history.rolling(5).mean()
df['MA10'] = history.rolling(10).mean()
df = df.dropna(subset=['pre_date_flow', 'MA5', 'MA10', 'p_flow'])
X = df[['pre_date_flow', 'MA5', 'MA10']]
y = df['p_flow']
split = int(len(X) * 0.8)
X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]
print('Random forest train/test:', len(X_train), len(X_test))
```

```python
random_forest_regressor = RandomForestRegressor(n_estimators=15, random_state=42)
# 拟合模型
random_forest_regressor.fit(X_train, y_train)
score = random_forest_regressor.score(X_test, y_test)
result = random_forest_regressor.predict(X_test)
print(f"Random forest test R²: {score:.6f}")

plt.figure(figsize = (15, 5))
plt.title('The result of Prediction')
plt.plot(y_test.to_numpy(), label='real')
plt.plot(result, label = 'predict')
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn import metrics

# 导入北京地铁站点15min进站客流
df = pd.read_csv('./in_15min.csv', encoding="gbk", parse_dates=True)
len(df)
df.dropna(inplace=True) # 首先去除空值所在的行

x_data = df.drop('Station_name', axis=1) # 以列为单位，删除'Station_name'列
```

```python
# 肘方法看K值
d = []
for i in range(1, 15):
    km = KMeans(n_clusters=i, init='k-means++', n_init=10, max_iter=300, random_state=0)
    km.fit(x_data)
    d.append(km.inertia_) # inertia簇内误差平方和

plt.plot(range(1, 15), d, marker='o')
plt.xlabel('number of clusters')
plt.ylabel('distortions')
plt.show()
```

```python
# 从效果图可以看出，K取6最合适
model_kmeans = KMeans(n_clusters=6, random_state=0, n_init=10)
model_kmeans.fit(x_data)
y_pre = model_kmeans.predict(x_data)
y_pre += 1
print('聚类结果为：' + str(y_pre))

# 评价指标
silhouette_s = metrics.silhouette_score(x_data, y_pre, metric = 'euclidean')
calinski_harabaz_s = metrics.calinski_harabasz_score(x_data, y_pre)
print('轮廓系数为：%.2f，c&h得分为：%d' % (silhouette_s, calinski_harabaz_s))
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.random.randn(100)
fig = plt.figure(figsize=(6, 4), dpi=50)  # 初始化一张画布，可以设置figsize和dpi
plt.plot(x)  # 直接在一张大的画布中画图，相当于获取当前活跃的的axes然后在上面作图
plt.show()
```

```python
fig = plt.figure(figsize=(8, 8), dpi=50)
ax = fig.add_subplot(2, 2, 1) # 在画布中添加一个axes(可以理解为子区域)，并返回这个子区域，参数的前两个表示子区域的行列数，最后一个表示子区域的顺序
ax.plot(x)   # 在该子区域中画图
ax2 = fig.add_subplot(2, 2, 2)
ax2.plot(x + 10)

# 另外一种写法：
fig = plt.figure(figsize=None, dpi=None)
fig.add_subplot(2, 2, 1)
plt.plot(x)   #获取当前活跃的的axes然后在上面作图
```
