```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-10,10)
y_sigmoid = 1/(1+np.exp(-x))
y_tanh = (np.exp(x)-np.exp(-x))/(np.exp(x)+np.exp(-x))

fig = plt.figure()
# plot sigmoid
plt.plot(x,y_sigmoid)
plt.grid()
plt.title('Sigmoid')
plt.show()
```

```python
import torch

input_x = torch.tensor([-2., -0.5, 0.5, 2.])
manual = 1 / (1 + torch.exp(-input_x))
print(manual)
```

```python
output = torch.sigmoid(input_x)
torch.testing.assert_close(manual, output)
print("Sigmoid matches:", output)
```

```python
import torch

input_x = torch.tensor([-2., -0.5, 0.5, 2.])
manual = torch.where(input_x >= 0, input_x, torch.zeros_like(input_x))
print(manual)
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-10,10)
y_sigmoid = 1/(1+np.exp(-x))
y_tanh = (np.exp(x)-np.exp(-x))/(np.exp(x)+np.exp(-x))

fig = plt.figure()
ax = fig.add_subplot()
y_relu = np.array([0.01*item  if item<0 else item for item in x ])
ax.plot(x,y_relu)
ax.grid()
ax.set_title('Leaky ReLu')
plt.show()
```

```python
import torch

input_x = torch.tensor([-2., -0.5, 0.5, 2.])
gamma = 0.01
manual = torch.where(input_x >= 0, input_x, gamma * input_x)
print(manual)
```

```python
output = torch.nn.functional.leaky_relu(input_x, negative_slope=gamma)
torch.testing.assert_close(manual, output)
print("Leaky ReLU matches:", output)
```
